```python
from transformers import pipeline
```

```python

ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)
```

```python
#inpout#
text = """
Hi i am najmun nahar i completed my B.Sc in CSE from BAUST in 2022.
"""
```

```python
#prediction#
result = ner(text)
```

```python
#output#
for entity in result:
    print(entity)
```

```python
# DataFrame#
import pandas as pd

df = pd.DataFrame(result)

df
```

```python
#csv#
df.to_csv("output.csv", index=False)
```
